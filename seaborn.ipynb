```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
# scatter plot -> axes level function
sns.scatterplot(data=tips, x = 'tip', y = 'total_bill', hue='day')
```

```python
# hue 
sns.scatterplot(data=tips, x = 'tip', y = 'total_bill', hue='sex')
```

```python
# style
sns.scatterplot(data=tips, x = 'tip', y = 'total_bill', hue='sex', style='day')
```

```python
tips.head()
```

```python
# size 
sns.scatterplot(data=tips, x = 'tip', y = 'total_bill', hue='sex', style='day', size='size')
```

```python
sns.relplot(kind='scatter', data=tips, x = 'tip', y = 'total_bill', hue='sex', style='day', size='size')
```

```python
# Line plot 
import plotly.express as px
gap  = px.data.gapminder()

gap
```

```python
gap['country'].unique()
```

```python
temp_df = gap[gap['country'].isin(['India', 'Germany', 'Brazil', 'Pakistan'])]
temp_df.sample(5)
```

```python
sns.relplot(kind='line', data=temp_df, x = 'year', y = 'gdpPercap', hue='country', style='continent')
```
