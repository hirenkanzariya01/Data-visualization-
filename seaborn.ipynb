```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.head()
```

```python
# scatter plot -> axes level function
sns.scatterplot(data=tips, x = 'total_bill', y = 'tip', hue='time', style='sex', size='size')
```

```python
sns.relplot(data=tips, x= 'total_bill', y='tip',  hue='time', style='sex', size='size')
```

```python
# Line plot 
import plotly.express as px
gap  = px.data.gapminder()
gap.shape
```

```python
temp_df = gap[gap['country'].isin(['India', 'Germany', 'Brazil', 'Pakistan'])]
temp_df.sample(5)
```

```python
sns.lineplot(data=temp_df, x = 'year', y = 'pop', hue='country')
```
