## 2. Matrix Plot
- Heatmap
- Clustermap

```python
import plotly.express as px
gap  = px.data.gapminder()
gap.head()
```

```python
gap['pop'] = round(gap['pop'] / 1000000)
gap.head()
```

```python
temp_df = gap.pivot(columns='year', values='pop', index='country').sample(10)
```

```python
# heatmap
# Plot rectangular data as a color-encoded matrix
plt.figure(figsize=(15,10))
sns.heatmap(temp_df,annot=True ,linewidths=2)
```

### Boxplot
- A boxplot is a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile [Q1], median, third quartile [Q3] and “maximum”). 
- It can tell you about your outliers and what their values are.
- Boxplots can also tell you if your data is symmetrical, how tightly your data is grouped and if and how your data is skewed

```python
# Box plot
tips = sns.load_dataset('tips')
tips.head()
```

```python
sns.boxplot(data=tips, x = 'day', y='total_bill', hue='smoker')
```
