```python
# import library 
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import warnings
warnings.filterwarnings('ignore')
```

 # Colored Scatterplots

```python
df = pd.read_csv('Datasets/iris.csv')
df.sample(5)
```

```python
df['Species'] = df['Species'].replace({"Iris-virginica":0,"Iris-versicolor":1, "Iris-setosa":2 })
```

```python
df.head()
```

```python
# cmap and alpha 
plt.scatter(df['SepalLengthCm'], df['PetalLengthCm'], c=df['Species'], alpha=0.9)
```

```python
# colorbar
plt.figure(figsize=(18, 10))
plt.scatter(df['SepalLengthCm'], df['PetalLengthCm'], c=df['Species'],  linewidth=2, linestyle="solid",  s=50)
plt.colorbar()
```

```python
# Plot size
plt.figure(figsize=(15, 10))
plt.scatter(df['SepalLengthCm'], df['PetalLengthCm'], c=df['Species'], cmap='winter')
plt.colorbar()
```

```python
# Annotations  use 
batters = pd.read_csv('Datasets/batter.csv')
newdf = batters.sample(25)
```

```python
newdf.head()
```

```python
newdf.iloc[3]['runs']
```

```python
plt.figure(figsize=(18, 10))
plt.scatter(newdf['avg'], newdf['runs'])
plt.title('Players Avg And Runs')
plt.xlabel('AVG')
plt.ylabel('RUNS')
for i in range(25):
    plt.text(newdf.iloc[i]['avg'], newdf.iloc[i]['runs'], newdf.iloc[i]['batter'], fontdict={"size":15})
```

```python
x = [1,2,3,4,5]
y = [10,20,25,20,30]
```

```python
# A Diff way to plot graph (subplots, add_subplots)
fig , ax = plt.subplots(ncols=3, nrows=2, figsize = (12,8) , sharex=True, sharey=True)
ax[0, 0].plot(x , y)
ax[0, 1].plot(x , y)
ax[1, 0].scatter(x , y)
ax[1, 1].bar(x, y)
ax[0, 2].scatter(x , y)
ax[1, 2].bar(x, y)

ax[0, 0].set_title('income')
ax[0,0].set_xlabel('Months')
ax[0,0].set_ylabel('income')

ax[0,1].set_title('income')
ax[0,1].set_xlabel('Months')
ax[0,1].set_ylabel('income')
plt.show()
```

```python
fig = plt.figure()
ax1 = fig.add_subplot(2,2,1)
ax1.plot(x , y)

ax2 = fig.add_subplot(2,2, 2)
ax2.pie(x)

ax3 = fig.add_subplot(2,2, 3)
ax3.bar(x , y)

ax4 = fig.add_subplot(2,2, 4)

```

```python
# 3d scatterploats
x = [2,5,10,20]
y = [0,3,16,4]
z = [0,12,4,16]



fig = plt.figure()
fig.add_subplot(projection='3d')
plt.plot(x , y , z)
plt.xlabel('---X---')
plt.ylabel('---Y---')
plt.title('3D Ploat')
# plt.zlabel('---Z---')
```
