```python
# import library 
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import warnings
warnings.filterwarnings('ignore')
```

 # Colored Scatterplots

```python
df = pd.read_csv('Datasets/iris.csv')
df.sample(5)
```

```python
df['Species'] = df['Species'].replace({"Iris-virginica":0,"Iris-versicolor":1, "Iris-setosa":2 })
```

```python
df.head()
```

```python
# cmap and alpha 
plt.scatter(df['SepalLengthCm'], df['PetalLengthCm'], c=df['Species'], alpha=0.9)
```

```python
# colorbar
plt.figure(figsize=(18, 10))
plt.scatter(df['SepalLengthCm'], df['PetalLengthCm'], c=df['Species'],  linewidth=2, linestyle="solid",  s=50)
plt.colorbar()
```

```python
# Plot size
plt.figure(figsize=(15, 10))
plt.scatter(df['SepalLengthCm'], df['PetalLengthCm'], c=df['Species'], cmap='winter')
plt.colorbar()
```

```python
# Annotations  use 
batters = pd.read_csv('Datasets/batter.csv')
newdf = batters.sample(25)
```

```python
newdf.head()
```

```python
newdf.iloc[3]['runs']
```

```python
plt.figure(figsize=(18, 10))
plt.scatter(newdf['avg'], newdf['runs'])
plt.title('Players Avg And Runs')
plt.xlabel('AVG')
plt.ylabel('RUNS')
for i in range(25):
    plt.text(newdf.iloc[i]['avg'], newdf.iloc[i]['runs'], newdf.iloc[i]['batter'], fontdict={"size":15})
```

```python
x = [1,2,3,4]
y = [5,6,7,8]
plt.scatter(x , y)
plt.text(1, 5, "point 1", fontdict={'color':'red', 'size':12})
plt.text(2, 6, "point 2")
plt.text(3, 7, "point 3")
plt.text(4, 8, "point 4")
```

```python
# Horizontal and Vertical lines
plt.figure(figsize=(10, 6))
plt.scatter(df['SepalLengthCm'], df['PetalLengthCm'], c=df['Species'], cmap='winter')
plt.colorbar()
plt.title('irs Datasets')
plt.xlabel('PetalLengthCm')
plt.ylabel('SepalLengthCm')
plt.axhline(5, color="red")
plt.axvline(7,color="green" )
plt.show()
```
