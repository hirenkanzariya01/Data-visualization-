```python
# Plot a bar chart of Units_Sold for each Model.
plt.bar(df['Model'], df['Units_Sold'])
plt.xlabel('--- Model ---')
plt.ylabel('--- Sold Units ---')
plt.title('model wise sold unit')
plt.ylim(15000, 16000)
plt.grid()
```

```python
df.head()
```

```python
grouped_revenue = df.groupby('Year')['Revenue_EUR'].sum()
```

```python
# Create a pie chart showing percentage share of Revenue_EUR.
plt.pie(grouped_revenue, labels=grouped_revenue.index, autopct='%0.1f%%')
plt.show()
```

```python
df.head()
```

```python
# Draw a line chart of Avg_Price_EUR for all models.
tempData = df.groupby('Model')['Avg_Price_EUR'].mean()
plt.plot(tempData.index,tempData)
plt.scatter(tempData.index , tempData)
plt.show()
```

```python
# Plot a histogram of Units_Sold.
plt.figure(figsize=(10,8))
plt.hist(df['Units_Sold'], bins=[1000,2000,3000,4000,5000,6000,7000,8000,9000,10000,11000,12000,13000,14000,15000,16000])
plt.xlabel('Sold Unit Count')
plt.grid()
```

```python
df.head()
df= df.sample(200)
```

```python
# Create a scatter plot between Avg_Price_EUR and Units_Sold.
plt.scatter(df['Avg_Price_EUR'], df['Units_Sold'], alpha=0.5, c=df['Year'])
plt.colorbar()
plt.xlabel("Avg_Price_EUR")
plt.ylabel("Units_Sold")
```

```python
# Plot Revenue_EUR vs Avg_Price_EUR and analyze correlation


Revenue_EUR_group = df.groupby('Year')['Revenue_EUR'].mean()
Avg_Price_EUR_group = df.groupby('Year')['Avg_Price_EUR'].mean()

fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(12,8))
ax[0].plot(Revenue_EUR_group.index, Revenue_EUR_group )
ax[0].set_xlabel('year')
ax[0].set_ylabel('Revenue_EUR_group')

ax[1].plot(Avg_Price_EUR_group.index, Avg_Price_EUR_group )
ax[1].set_xlabel('year')
ax[1].set_ylabel('Avg_Price_EUR_group')

```

```python
corr = np.corrcoef(df['Revenue_EUR'], df['Avg_Price_EUR'])
```
