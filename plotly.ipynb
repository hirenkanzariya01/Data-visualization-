```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
import plotly.express as px
```

```python
# • Scatter Plot
df = px.data.tips()
fig = px.scatter(df, x='tip', y='total_bill', color='day')
fig.show()
```

```python
# • Line Chart
df = px.data.gapminder().query("country == 'India' or country == 'Pakistan'")
fig = px.line(df, x='year', y='pop')
fig.show()
```

```python
df.sample(5)
```

```python
# • Bar Chart
px.bar(df, x='year', y='pop', color='country')
```

```python
# • Histogram
px.histogram(df, x='year', y='gdpPercap')
```

```python
# • Box Plot
df = px.data.gapminder().query("country == 'India' or country == 'Pakistan'or country == 'China'")
px.box(df, x = 'year', y='pop')
```

```python
# • violin
df = px.data.tips()
px.violin(df , x = 'day', y='total_bill', title='Total Bill' )
```

```python
# • Pie Chart
df = pd.read_csv('gayle-175.csv')
px.pie(df , values='batsman_runs', names='batsman', title="total Runs ")
```

```python
# • Donut Chart
px.pie(df, values='batsman_runs', hole=0.5)
```
